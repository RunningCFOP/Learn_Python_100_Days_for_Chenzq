## 函数和模块
***
在进行本节课的学习之前，先尝试解决如下数学问题，请说出下面的方程组有多少组正整数解。
$$
x_1 + x_2 + x_3 + x_4 = 8
$$
这个问题与"将8个苹果分成四组且每组至少一个苹果有多少种方案"，也等价于分隔8个苹果的7个间隙之间放入三个隔断将苹果分成四组有多少种方案，所以答案是$C^3_7 = 35$，其中$C^3_7$代表7选3的组合数，其计算公式如下所式。
$$
C^n_m = \frac{m!}{n!(m - n) !}
$$
根据之前学习的知识，可以用循环做出累乘的方式分别计算出$m!$、$n!$和$(m - n)!$，然后不规则通过除尘运算得到组合数$C^n_m$,代码如下所示。

```python
# 输入m和 n计算组合数C(m, n)的值
m = int(input('请输入m = '))
n = int(input('请输入n = '))

times_m, times_n, times_mn = 1, 1, 1

for i in range(1,m + 1):
    times_m *= i
for j in range(1, n + 1):
    times_n *= j
for k in range(1, m - n + 1):
    times_mn *= k

c_mn = times_m / (times_n * times_mn)
print(f'{c_mn =: .0f}')
```

上在的代码中，我们做了三次求阶乘的操作，虽然m、n、m - n的值各不相同，但是三段代码并没有实质性区别，属于重复代码。世界级的编程大量Martin Fowler曾说过：“**代码有很多种坏味道，重复是最坏的一种！**”。要写出高质量的代码，首先就要解决重复代码的问题。对于上述的代码来说，可以将求阶乘的功能封装到一个称为“函数”的代码块中，在需要计算阶乘的地方，只需要“调用函数”即可实现对求阶乘功能的复用。

## 定义函数

数学上的函数通常形如$y = f(x)$或者$z = g(x, y)$这样的形式，在$y = f(x)$中，$f$是函数的名字，$x$是自变量，$y$是函数的因变量，而在$z = g(x, y)$中，$g$是函数名,$x$和$y$是自变量，$z$是函数的因变量。Python中的函数跟这个结构是一致的，**每个函数都有自己的名字、自变量和因变量**。通常把Python函数的自变量称为**函数的参数**，而因变量称为函数的**返回值**。

Python中可以使用`def`关键字来定义函数，和变量一样每个函数都有一个名字，命名规则和变量的命名规则是一样的。在函数名后面的圆括号可以设置函数的参数，即函数的自变量，而函数执行完成后，会通过`returen`关键字来返回函数的执行结果，就函数的因变量。如果函数中没有`return`语句，那么函数会返回代表空值的`None`。 另外，函数也可以没有自变量，但是函数后面的圆括号是必须有的。

一个函数要做的事情（要执行的代码），是通过代码缩进的方式放大函数定义行之后，和之前的分支和循环结构的代码块类似，如下图所示：
![funciton_image](../image/function_definition.png)

接下来，将之前求阶乘的代码操作入到一个函数中，通过这种方式来重构上面的代码。**所谓重构，是在不影响代码执行结果的前提下对代码的结构进行调整**，重构之后的代码如下所示。

```python
# 通过关键字def定义求阶乘的函数
# 自变量（参数）num是一个非负整数
# 因变量（返回值） 是num的阶乘
def fac(num):
    result = 1
    for n in range(2, num + 1):
        result *= n
    return result

m = int(input('m = '))
n = int(input('n = '))

# 计算阶乘的时候不需要写重复的代码而是直接调用函数
# 调用函数的语法是在函数名后面跟上圆括号并传入参数

print(fac(m) // fac(n) // fac(m - n))
```

上面的代码更加的简洁和优雅，再重要是的，所定义的求阶乘函数`fac`还可以在其他需要求阶乘的代码中重复使用。所以，**使用函数可以将功能上相对独立且会被重复使用的代码封装起来**，当需要使用这些代码时，不是把重复的代码再编写一遍，而是**通过调用函数实现对即有代码的复用**。事实上，Python的标准库的`math`中，已经有一个名为`factorial`的函数实现了求阶乘的功能，可以直接用`import math`导入`math`模块，然后使用`math.factorial`来调用求阶乘的函数，也可以通过`from math import factorial`直接导入`factorial`函数来使用它，代码如下所示。

```python
from math import factorial

m = int(input('m = '))
n = int(input('n = '))

print(factorial(m) // factorial(n) // factorial(m - n))
```

如果标准的函数名称太长， 也可以使用`as`关键字为其别名。

```python
from math import factorial as f

m = int(input('m =  '))
n = int(input('n = '))

print(f(m) // f(n) // f(m - n))
```

## 函数的参数
### 位置参数和关键字参数
再以一个函数为示例，根据给出的三条边的长度判断是否可以构成三角形，如果可以构成则返回`True`，否则返回`False`，代码如下：

```python
def make_judgement(a, b, c):
    """
    判断三条边长能否构成一个三角形
    :param a: 边长
    :param b:
    :param c:
    :return: bool值，可以为True，否则为False
    """
    return a + b > c and a + c > b and b + c > a
```

上面的`make_judgement`函数有三个参数，这种参数叫做**位置参数**，在调用函数时通常按照从左到右的顺序依次传入，而传入参数的数量必须和定义函数时的参数的数量相同，如下所示：

```python
# 运行本单元前要先运行上一个单元
print(make_judgement(1, 2, 3)) # False
print(make_judgement(3, 4, 5)) # True
```

如果不想按照从左到右的顺序给出`a`、`b`、`c`三个参数的值，也可以使用关键字参数，通过“参数名 = 参数值“的形式为函数传入参数，如下面代码所示：

```python
print(make_judgement(b = 2, c = 3, a = 1)) # False
print(make_judgement(c = 6, b = 4, a = 5)) # True
```

在定义函数是，可以在参数列表中用`/`设置强制位置参数（*positional-only argument*），用`*`设置**命名关键字参数**。所谓强制位置参数，就是调用函数时，只能按照参数位置来接收参数值的参数；而命名关键字参数只能通过“参数名 = 参数值“的方式来传递和接收参数，如下面例子所示：

```python
# / 前面的的参数是强调位置参数
def make_judgement(a, b, c, /):
    """
    判断三条边的长度能否构成三角形
    :param a:
    :param b:
    :param c:
    :return:
    """
    return a + b > c and a + c > b and b + c > a
print(make_judgement(1, 2, 3)) # False

# 下面的两行代码都会产生TypeError错误，错误信息提示”强制位置参数是不允许给出参数名的“
# TypeError: make_judgement() got some positional-only arguments passed as keyword arguments: 'a, b, c'
print(make_judgement(a = 1, b = 2, c = 3))
print(make_judgement(b = 1, c = 2, a = 3))
```

> 说明：强制位置参数是Python3.8引入的特性。

```python
# * 后面的参数是命名关键字参数
def make_judgement(*, a, b, c):
    """
    判断三条边长能否构成三角形
    :param a:
    :param b:
    :param c:
    :return:
    """
    return a + b > c and a + c > b and b + c > a

print(make_judgement(a = 1, b = 2, c = 3))  # False
print(make_judgement(b = 1, a = 2, c = 3))  # False

# 下面的代码会产生TypeError错误，错误信息提示”函数没有位置参数，却给了3个位置参数“
# TypeError: make_judgement() takes 0 positional arguments but 3 were given
print(make_judgement(1, 2, 3))
```

### 参数的默认值
Python中允许函数的参数拥有默认值，可以将之前的CRAPS（day07）赌博游戏中的骰子获得点数功能封装到函数中，如下面的代码所示：

```python
from random import randrange

# 定义摇色子的函数
# 函数的自变量（参数）n表示色子的个数，默认值为2
# 函数的因变量（返回值）表示摇n颗色子得到的点数
def roll_dice(n = 2):
    total = 0
    for _ in range(n):
        total += randrange(1, 7)
    return total

# 如果没有指定参数，那么使用默认值2，表示摇两颗色子
print(roll_dice())

# 如果传入参数3，变量n被赋值为3，表示摇三颗色子获得点数
print(roll_dice(3))
```

```python
# 下面是一个更简单的例子
def add(a = 0, b = 0, c = 0):
    """三个数相加求和"""
    return a + b + c

# 调用add函数，没有传入参数，那么a，b，c都默认值为0
print(add()) # 0
# 调用add函数，传入一个参数，该参数赋值给变量a，变量b和变量c仍然为默认值0
print(add(1)) # 1
# 调用add函数，传入两个参数，分别赋值给变量a和变量b，变量c使用默认值0
print(add(1, 2)) # 3
# 调用add函数，传入三个参数，分别赋值给a、b、c三个变量
print(add(1, 2, 3)) # 6
```

需要注意的是，**带默认值的参数必须放在不带默认值的参数之后**，否则将产生`SyntaxError`,错误消息是`SyntaxError: non-default argument follows default argument`,即”没有默认值的参数放在了带默认值的参数后面“。

```python
def add(a = 0, b, c):
    """三个数相加求和"""
    return a + b + c
```

### 可变参数
Python语言中可以通过星号表达式语法让函数支持可变参数。所谓**可变参数**指的是在调用函数时，可以向函数传入`0`个或任意多个参数。下面的代码演示了如何使用可变位置参数实现对任意多个数求和的`add`函数，调用函数时传入的参数会保存到一个元组，通过对该元组的遍历，可以获取传入函数的参数。

```python
# 用星号表达式还表示args可以接收0个或多个参数
# 调用函数时传入n个参数会组装成一个n元组赋给args
# 如果一个参数都没有传入，那么args会是一个空元组
def add(*args):
    total = 0
    # 对保存的变量进行遍历
    for val in args:
        # 对参数进行类型检查（数值型才能求和）
        if type(val) in (int, float):
            total += val
    return total

# 在调用add函数时，可以传入0个或任意多个参数
print(add())
print(add(1))
print(add(1, 2, 3))
print(add(1, 2, 'hello', 3.45, 6)) # 12.45
```

如果希望通过”参数名 = 参数值”的形式传入若干个参数，具体有多少个参数也是不确定的，可以通过给添加可变关键字参数，把传入的关键字参数组装到一个字典中，代码如下所示：

```python
# 参数列表中的**kwargs可以接收0个或任意多个关键字参数
# 调用时传入的关键字参数会组装成一个字典（参数名是字典中的键，参数值是字典中的值）
# 如果一个关键字参数都没有传入，那么kwargs会是一个空字典。
def foo(*args, **kwargs):
    print(args)
    print(kwargs)

foo(3, 2.1, True, name = 'chenzq', age = 43, gpa = 4.95, signatrue = [])
```

### 用模块管理函数
给变量、函数取名字都是一个问题，总会遇到**命名冲突**这种尴尬的情况。最简单的场景就是在同一个`.py`文件中定义了两个同各函数，如下所示。

```python
def foo():
    print('hell, world!')

def foo():
    print('goodbye, world!')

foo() # goodbye, world!
```

如何避免上述冲突呢？因为Python中每个文件就代表了一个模块（module），在不同模块中可以有同名的函数，在使用函数时，通过`import`关键字导入指定的模块再使用**完全限定名**（`模块名.函数名`）的调用方式，就可以区分到底要使用哪个模块中的`foo`函数，代码如下所示(已经定义好了两个module1/2文件):

```python
import sys
sys.path.append('./day1-20/res/day14') # 添加相对路径

import module1
import module2

module1.foo() # hello world!
module2.foo() # goodbye, world!
```

当然，在导入模块时，仍可以使用`as`关键词对模块进行别名，可使用更为简短的**完成限定名**。

```python
from module1 import foo as f1
from module2 import foo as f2
f1()
f2()
```

但是，如果从不同的中导入了同名的函数，后面导入的函数会替换掉之前的导入，如下面的代码所示，调用`foo`会输出`goodbye，world！`，因为先导入的是`module1`的`foo`，后导入了`module2`的`foo`，如果两个`from...import...`反过来，则输入相反。

```python
from module1 import foo
from module2 import foo
foo() # goodbye, world!
```
